```python
import pandas as pd
from functools import reduce
from sklearn.preprocessing import MinMaxScaler

# --- 1. 定義檔案路徑佔位符 (Placeholder for File Paths) ---
# 將所有實際的本地檔案路徑替換成佔位符，以便上傳到GitHub。

SPORT_EXCEL_PATH = r'path/to/Sport_index/sport_index.xlsx'
DAYTIME_ENTERTAINMENT_EXCEL_PATH = r'path/to/DayTime_entertainment_index/Daytime_Entertainment_index.xlsx'
_24HRS_EXCEL_PATH = r'path/to/24小時便利度index/24hrs_index.xlsx'
OPENRICE_EXCEL_PATH = r'path/to/openRice_index/openrice_index.xlsx'
ART_EXCEL_PATH = r'path/to/Art/Art_index.xlsx'
ACCESSIBILITY_EXCEL_PATH = r'path/to/Accessibility/travel_time.xlsx'
OUTPUT_EXCEL_PATH = r'path/to/output/normalized_database.xlsx' # 輸出路徑也替換

# --- 2. 載入並清理各個數據框 (Load and Clean DataFrames) ---

# Sport
Sport_df = pd.read_excel(SPORT_EXCEL_PATH).iloc[:, [0, -2, -1]]
Sport_df.columns = ['District', 'Sport_total', 'Sport_Diversity']

# Daytime Entertainment
daytime_entertainment_df = pd.read_excel(DAYTIME_ENTERTAINMENT_EXCEL_PATH)
daytime_entertainment_df.columns = ['District', 'daytime_entertainment_Malls', 'daytime_entertainment_Movie']

# 24hrs Convenience
_24hrs_df = pd.read_excel(_24HRS_EXCEL_PATH).iloc[:, [0, 1]]
_24hrs_df.columns.values[-1] = '24hrs'

# Openrice
openrice_df = pd.read_excel(OPENRICE_EXCEL_PATH).iloc[:, [0, -1]]
openrice_df.columns.values[-1] = 'openrice'

# Art
Art_df = pd.read_excel(ART_EXCEL_PATH).iloc[:, [0, -2]]
Art_df.columns.values[-1] = 'Art'

# Accessibility
Accessibility_df = pd.read_excel(ACCESSIBILITY_EXCEL_PATH)
Accessibility_df.columns = ['District', 'To_Central', 'To_KwunTong']

# --- 3. 合併數據框 (Merge DataFrames) ---
df_merged = reduce(
    lambda left, right: pd.merge(left, right, on='District'), 
    [Sport_df, daytime_entertainment_df, _24hrs_df, openrice_df, Art_df, Accessibility_df]
)

print("--- Merged Data Head ---")
print(df_merged.head())

# --- 4. 數據正規化 (Data Normalization) ---

# Keep the 'District' column separately
district_col = df_merged[['District']]

# Select numeric columns only
numeric_cols = df_merged.columns[1:]
numeric_data = df_merged[numeric_cols]

# Normalize using MinMaxScaler
scaler = MinMaxScaler()
normalized_data = scaler.fit_transform(numeric_data)

# Convert back to DataFrame
normalized_df = pd.DataFrame(normalized_data, columns=numeric_cols)

# Concatenate the District column back
final_df = pd.concat([district_col, normalized_df], axis=1)

print("\n--- Normalized Data Head (First 5 Rows) ---")
print(final_df.head())

# --- 5. 輸出正規化後的數據 (Output Normalized Data) ---
# 將輸出路徑也替換為佔位符
final_df.to_excel(OUTPUT_EXCEL_PATH, index=False)
```
